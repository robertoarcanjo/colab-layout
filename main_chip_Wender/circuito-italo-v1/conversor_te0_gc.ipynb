# Conversor TE0–TE0 com grating couplers

Os guias de 0,5 µm e de 1,1 µm compartilham a borda de baixo. Os centros diferem 0,3 µm e, com um conversor virado 180°, as portas TE0 ficam 0,6 µm uma acima da outra. Cada acoplador respeita os 12 µm da sua porta; a distância entre os dois sai **24,6 µm**.
